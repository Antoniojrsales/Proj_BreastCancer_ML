- Analise estrutural

```python
print("-" * 25)
print("Dimensões do DataFrame:")
print("-" * 25)
print(f"Linhas: \t {df_cancer.shape[0]}")
print(f"Colunas:\t {df_cancer.shape[1]}")
```

```python
print('-'*30)
print('Verificando se a dados nulos')
print('-'*30)
df_cancer.isnull().sum()
```

```python
print('-'*35)
print('Verificando se a dados duplicados')
print('-'*35)
df_cancer.duplicated().sum()
```

```python
print('-'*50)
print('Verificando se a dados duplicados na coluna ID')
print('-'*50)
df_cancer.index[df_cancer.index.duplicated()]
```

```python
print('-'*50)
print('Intera sobre as colunas e verifica se a unicidade')
print('-'*50)
for coluna in df_cancer.columns:
    num_unicos = df_cancer[coluna].nunique()
    num_linhas = df_cancer.shape[0]
    if num_unicos == num_linhas:
        print(f"A coluna '{coluna}' possui valores únicos.")
    else:
        print(f"A coluna '{coluna}' não possui valores únicos.")
```

```python
print("-" * 40)
print("Contagem de valores únicos por coluna:")
print("-" * 40)
print(df_cancer.nunique())
```

```python
df_cancer.describe().style.set_caption("Resumo estatistico dos dados")
```

```python
print('-'*25)
print('Contagem por amostras (malignas (0) e benignas (1))')
print('-'*25)
agrupando_target = df_cancer.groupby(['target']).size()
agrupando_target.index = ['Benigno (0)', 'Maligno (1)']
agrupando_target
```

```python
print('-'*30)
print("Contagem por amostras em %:")
print('-'*30)
agrupando_target = df_cancer['target'].value_counts(normalize=True).mul(100).round(2).astype(str) + '%'
agrupando_target.index = ['Benigno (0)', 'Maligno (1)']
agrupando_target
```
