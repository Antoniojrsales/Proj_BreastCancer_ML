```python
import pandas as pd
from sklearn.datasets import load_breast_cancer
import plotly.express as px
import numpy as np
```

```python
cancer = load_breast_cancer()
x_cancer = cancer.data
y_cancer = cancer.target
```

```python
x_cancer = pd.DataFrame(x_cancer, columns=cancer['feature_names'])
pd.set_option('display.max_columns', None)
x_cancer['target'] = y_cancer
df_cancer = x_cancer
```

```python
agrupando_target = df_cancer.groupby(['target']).size()
agrupando_target.index = ['Benigno (0)', 'Maligno (1)']
contagem_percentual = agrupando_target.values / agrupando_target.values.sum() * 100
texto_completo = [
    f'{contagem} ({perc:.2f}%)'
    for contagem, perc in zip(agrupando_target.values, contagem_percentual)
]


fig = px.bar(
    x=agrupando_target.index, 
    y=agrupando_target.values,
    title='Distribuição de Amostras (Benigno vs Maligno)', # Define o título principal do gráfico
    labels={'x': 'Diagnóstico', 'y': 'Número de Amostras'}, # Define os títulos dos eixos X e Y
    color=agrupando_target.index, # Usa os nomes das categorias para definir as cores das barras
    color_discrete_sequence=['#1f77b4', '#d62728'], # Define as cores específicas (Vermelho e Azul) para cada categoria
    text=texto_completo, # Usa os valores (contagens, porcentagem) como texto flutuante sobre as barras
    height=500 # Define a altura total do gráfico em pixels
)
fig.update_traces(
    textfont_size=14, # Define o tamanho da fonte dos números (rótulos de texto)
    textangle=0, # Mantém o texto (os números) sem rotação (0 graus)
    textposition="outside", # Posiciona o texto FORA e acima das barras
    cliponaxis=False, # Permite que o texto 'outside' não seja cortado pelo limite do gráfico
    marker_line_width=1, # **SUGESTÃO 1:** Define a largura da borda de cada barra
    marker_line_color='black' # **SUGESTÃO 1:** Define a cor da borda de cada barra como preta
)

fig.update_layout(
    plot_bgcolor='rgba(0,0,0,0)', # Torna o fundo da área do gráfico (onde as barras estão) transparente
    paper_bgcolor='lightgray', # **SUGESTÃO 2:** Define a cor do fundo da página/janela como branco para um visual mais limpo
    title_x=0.5, # Centraliza o título principal horizontalmente (0.5 = 50% da largura)
    title_font_size=22, # Use o valor que desejar (padrão é geralmente 18)
    showlegend=False # **SUGESTÃO 3:** Oculta a legenda, pois as cores já estão explicadas no eixo X                
)

fig.update_xaxes(tickfont_size=14) # Define o tamanho da fonte dos ticks do eixo X
fig.update_yaxes(tickfont_size=12, automargin=True) # Ajusta automaticamente a margem do eixo Y para evitar cortes de rótulos (melhor prática)
fig.show()
```

```python
fig = px.histogram(
    df_cancer,
    x='mean radius',
    color='target',
    color_discrete_map={0:'#1f77b4', 1:'#d62728'},
    opacity=0.6, # Torna as barras semi-transparentes para ver a sobreposição (NOVO/AJUSTADO)
    nbins=20, # Número de intervalos (bins)
    histnorm='probability density', # Normaliza o eixo Y para densidade (para comparar formas) (NOVO)
    marginal='box', # Adiciona um Box Plot na margem para resumir estatísticas (NOVO)
    title='Distribuição do Raio Médio por Diagnóstico', # Define o título principal do gráfico
    height=720
)

fig.update_traces(
    marker_line_width=1, # **SUGESTÃO 1:** Define a largura da borda de cada barra
    marker_line_color='black' # **SUGESTÃO 1:** Define a cor da borda de cada barra como preta
)

fig.update_layout(
    plot_bgcolor='rgba(0,0,0,0)', # Torna o fundo da área do gráfico (onde as barras estão) transparente
    paper_bgcolor='lightgray', # **SUGESTÃO 2:** Define a cor do fundo da página/janela como branco para um visual mais limpo
    title_x=0.5, # Centraliza o título principal horizontalmente (0.5 = 50% da largura)
    title_font_size=22, # Use o valor que desejar (padrão é geralmente 18)
    showlegend=False, # **SUGESTÃO 3:** Oculta a legenda, pois as cores já estão explicadas no eixo X
    barmode='overlay' # Garante que as distribuições se sobreponham (NOVO)             
)

fig.update_xaxes(tickfont_size=14, title_text='Raio Médio (mean_radius)') # Define o tamanho da fonte dos ticks do eixo X
fig.update_yaxes(tickfont_size=12, automargin=True, title_text='Densidade de Probabilidade') # Ajusta automaticamente a margem do eixo Y para evitar cortes de rótulos (melhor prática)
fig.show()
```

```python
fig = px.box(
    df_cancer,
    x=df_cancer['target'].map({0: 'Benigno', 1: 'Maligno'}),
    y='mean radius',
    color='target', # Usa a coluna 'target' (0 e 1) para o mapeamento de cores
    labels={
        'x': 'Diagnóstico', # Nome do Eixo X
        'y': 'Raio Médio',  # Nome do Eixo Y
        'color': 'Diagnóstico' # Nome da Legenda de Cores
    },
    title='Distribuição do Raio Médio por Diagnóstico',
    color_discrete_map={0:'#1f77b4', 1:'#d62728'},
    height=500
)

fig.update_layout(
    plot_bgcolor='rgba(0,0,0,0)', # Torna o fundo da área do gráfico (onde as barras estão) transparente
    paper_bgcolor='lightgray', # Define a cor do fundo da página/janela
    title_x=0.5, # Centraliza o título principal horizontalmente (0.5 = 50% da largura)
    title_font_size=22, # Use o valor que desejar (padrão é geralmente 18)
    showlegend=False # Oculta a legenda              
)

fig.update_xaxes(tickfont_size=14, title_text='Diagnóstico') # Define o tamanho da fonte dos ticks do eixo X
fig.update_yaxes(tickfont_size=12, automargin=True, title_text='Raio Médio (mean_radius)') # Ajusta automaticamente a margem do eixo Y para evitar cortes de rótulos
fig.show()
```

```python
#-Calcule a correlação entre as colunas.
correlacao = df_cancer.drop('target', axis=1).corr()
correlacao.head()
```

```python
# 3. Criar o Heatmap de Correlação
fig = px.imshow(
    correlacao,
    text_auto='.2f',  # Exibe os valores de correlação na célula
    aspect="auto",   # Ajusta o aspecto para melhor visualização
    color_continuous_scale='RdBu_r',  # Usa um esquema de cores divergente (vermelho/azul)
    range_color=[-1, 1],             # Garante que a escala de cores vá de -1 a 1 (correlação)
    title='Matriz de Correlação entre as Features do Dataset Breast Cancer',
    height=700

)

fig.update_layout(
    plot_bgcolor='rgba(0,0,0,0)', # Torna o fundo da área do gráfico (onde as barras estão) transparente
    paper_bgcolor='lightgray', # Define a cor do fundo da página/janela
    title_x=0.5, # Centraliza o título principal horizontalmente (0.5 = 50% da largura)
    title_font_size=22, # Use o valor que desejar (padrão é geralmente 18)  
    font=dict(size=12), # Reduzir ligeiramente o tamanho da fonte para encaixar mais rótulos       
)

fig.update_traces(
    textfont_size=10 # Aumenta o tamanho da fonte do texto na célula.
)

fig.update_xaxes(
    tickangle=45, # Rotação de 45 graus para os rótulos do eixo X
    automargin=True
)
fig.update_yaxes(
    automargin=True # Ajusta a margem do eixo Y automaticamente
)

fig.show()
```

```python
fig = px.scatter(
    df_cancer,
    x='mean radius',
    y='mean area',
    color='target',
    title='Analisando a Correlação entre Raio Médio e Área Média',
    trendline='ols', # Adicionar reta de tendência para visualizar a correlação
    color_discrete_map={0:'#1f77b4', 1:'#d62728'},
    labels={
        'target': 'Diagnóstico',
        'x': 'Raio Médio (mean radius)',
        'y': 'Área Média (mean area)'
    },
    height=600
)

# Ajustes para melhor visibilidade dos pontos
fig.update_traces(
    marker=dict(size=8, opacity=0.8, line=dict(width=1, color='DarkSlateGrey')),
    selector=dict(mode='markers')
)

fig.update_layout(
    plot_bgcolor='rgba(0,0,0,0)', # Torna o fundo da área do gráfico (onde as barras estão) transparente
    paper_bgcolor='lightgray', # Define a cor do fundo da página/janela
    title_x=0.5, # Centraliza o título principal horizontalmente (0.5 = 50% da largura)
    title_font_size=22, # Use o valor que desejar (padrão é geralmente 18)
    showlegend=True            
)

fig.update_xaxes(tickfont_size=14, title_text='Raio Médio (mean radius)') # Define o tamanho da fonte dos ticks do eixo X
fig.update_yaxes(tickfont_size=12, automargin=True, title_text='Área Média (mean area)') # Ajusta automaticamente a margem do eixo Y para evitar cortes de rótulos
fig.show()
```
