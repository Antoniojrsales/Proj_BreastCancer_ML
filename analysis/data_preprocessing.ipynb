```python
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import numpy as np
```

✅ - Realize a leitura do dataset breast_cancer

```python
cancer = load_breast_cancer()
x_cancer = cancer.data
y_cancer = cancer.target
```

✅ - Transformando os dados em um DataFrame

```python
x_cancer = pd.DataFrame(x_cancer, columns=cancer['feature_names'])
pd.set_option('display.max_columns', None)
x_cancer['target'] = y_cancer
df_cancer = x_cancer
```

```python
correlacao = df_cancer.drop('target', axis=1).corr()

# Cria uma cópia da matriz e zera a parte triangular superior e a diagonal
# Isso evita a duplicidade de pares (A vs B e B vs A) e elimina a correlação de 1.0 (A vs A)
upper = correlacao.where(np.triu(np.ones(correlacao.shape), k=1).astype(bool))

# Filtra a matriz pelos valores absolutos maiores que 0.85 (Correlação muito forte)
to_drop = [column for column in upper.columns if any(upper[column].abs() > 0.85)]

print("Pares de features com |r| > 0.85:")
print(upper[to_drop])
```

```python
# Lista de features redundantes (alta multicolinearidade) a serem removidas
features_to_drop = [
    # Grupo 1: Raio/Perímetro/Área (mantendo o raio como principal)
    'mean perimeter', 
    'mean area',
    'worst perimeter', 
    'worst area',
    'perimeter error', 
    'area error',
    
    # Grupo 2: Concavidade/Compactação (mantendo 'points')
    'mean concavity', 
    'worst concavity', 
    'concave points error', # Mantendo 'compactness error'
    
    # Outras correlações fortes
    'worst radius' # Altamente correlacionado com 'mean radius'
]

# 2. Criar o DataFrame Final Limpo (Incluindo a 'target' para facilitar a carga)
# Se df_cancer é o seu DataFrame original:
df_cleaned = df_cancer.drop(features_to_drop, axis=1)

# 3. Salvar o DataFrame em CSV
# Sugestão de nome:
caminho_arquivo = '../data/breast_cancer_cleaned.csv' 

df_cleaned.to_csv(caminho_arquivo, index=False) 

print(f"DataFrame limpo, com 19 features e a coluna 'target', salvo em: {caminho_arquivo}")
```
