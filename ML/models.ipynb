```python
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import numpy as np
```

```python
url = '../data/breast_cancer_cleaned.csv'
df_cancer = pd.read_csv(url)
df_cancer
```

```python
X = df_cancer.drop('target', axis=1)
y = df_cancer['target']

#Separando amostras de 70% para treino e 30% para teste#
X_train, X_test, y_train, y_test = train_test_split(X, y, 
                                                    test_size=0.2, 
                                                    stratify=y,
                                                    random_state=42, 
                                                    shuffle=True)
```

```python
# 1. Inicializar o Scaler
scaler = StandardScaler()

# 2. Treinar o Scaler (FIT) APENAS nos dados de TREINO.
# O scaler aprende a média e o desvio padrão do conjunto de treino.
scaler.fit(X_train)

# 3. Aplicar a Transformação (TRANSFORM) em AMBOS os conjuntos
# Aplicamos a transformação usando os parâmetros aprendidos no FIT
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Tamanho do conjunto de treinamento: {}".format(len(X_train_scaled)))
print("Tamanho do conjunto de testes: {}".format(len(X_test_scaled)))

# Verificação de Escalonamento (Obrigatório antes de Modelar!)
print("Média (deve ser próximo de zero):")
print(X_train_scaled.mean(axis=0))

print("\nDesvio Padrão (deve ser próximo de um):")
print(X_train_scaled.std(axis=0))
```
